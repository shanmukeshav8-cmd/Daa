# Airport assistant data exploration

Explore the bundled illustrative training data. These records do not describe live airport conditions.

```python
import json
from pathlib import Path

import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "data").is_dir():
    ROOT = ROOT.parent
delay = pd.read_csv(ROOT / "data" / "flight_delay.csv")
queue = pd.read_csv(ROOT / "data" / "queue_data.csv")
with (ROOT / "data" / "intents.json").open(encoding="utf-8") as file:
    intents = json.load(file)["intents"]
print(f"Project root: {ROOT}")
print(f"Delay rows and columns: {delay.shape}")
print(f"Queue rows and columns: {queue.shape}")
```

## Data quality and target summaries

```python
display(pd.DataFrame({"flight_delay_missing": delay.isna().sum(), "queue_missing": queue.isna().sum()}))
display(delay[["delay_minutes"]].describe().round(2))
display(queue[["queue_minutes"]].describe().round(2))
```

## Feature distributions

```python
for column in ["airline", "weather", "traffic_level"]:
    print(f"\nFlight delay — {column}")
    display(delay[column].value_counts().rename("records").to_frame())
print("\nQueue records by terminal")
display(queue["terminal"].value_counts().sort_index().rename("records").to_frame())
print("\nQueue records by hour")
display(queue["hour"].value_counts().sort_index().rename("records").to_frame())
```

## Intent coverage

```python
intent_counts = pd.DataFrame(
    [{"intent": item["tag"], "examples": len(item["patterns"])} for item in intents]
).sort_values("intent")
display(intent_counts.set_index("intent"))
```

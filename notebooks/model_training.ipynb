# Train airport assistant models

Run the same checked training scripts used by the command-line workflow. The bundled data is illustrative; resulting estimates are not live forecasts.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / "train_chatbot.py").is_file():
    ROOT = ROOT.parent
assert (ROOT / "train_chatbot.py").is_file(), f"Project root not found from {Path.cwd()}"
print(f"Project root: {ROOT}")
```

## Train intent classifier and regression estimators

```python
for script in ["train_chatbot.py", "train_delay.py", "train_queue.py"]:
    subprocess.run([sys.executable, str(ROOT / script)], cwd=ROOT, check=True)
```

## Confirm generated model files

```python
model_paths = [
    ROOT / "models" / "chatbot_model.pkl",
    ROOT / "models" / "delay_model.pkl",
    ROOT / "models" / "queue_model.pkl",
]
missing = [path for path in model_paths if not path.is_file()]
if missing:
    raise FileNotFoundError(f"Training did not produce: {missing}")
for path in model_paths:
    print(f"Ready: {path.relative_to(ROOT)}")
```
